# 01 Data Exploration
Synthetic dataset exploration for AquaGuard.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/synthetic_water_consumption_2026_90d_6b_seed42.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.head()
```

```python
df.groupby('building_name')['water_consumption_liters'].describe()[['mean','std','min','max']]
```

```python
sns.lineplot(data=df.groupby('hour',as_index=False)['water_consumption_liters'].mean(),x='hour',y='water_consumption_liters')
plt.title('Average consumption by hour')
plt.show()
```

```python
sns.boxplot(data=df, x='is_weekend', y='water_consumption_liters')
plt.title('Weekday vs weekend consumption')
plt.show()
```
